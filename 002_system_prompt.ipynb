```python
import os
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))
```

```python
def add_user_message(messages, text):
    user_message = {"role":"user", "content":text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content":text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=0.0):
    params = {
        "model" : "claude-sonnet-4-5-20250929",
        "max_tokens" : 1024,
        "messages": messages,
        "temperature": temperature,
    }

    if system:
        params["system"] = system
    message = client.messages.create(**params) 
    return message.content[0].text
```

```python
messages = []


system = """
    You are a senior python programmer. You write the single-best approach for a 
    given problem with the best time and space complexity and minimal explanation"""


add_user_message(messages, "Generate an niche idea for a qwen-cloud hackathon")
answer = chat(messages, temperature=1.0)#, system=system)
print(answer)

```

```python
add_assistant_message(messages, answer)
add_user_message(messages, "Since we need to isolate x maybe divide everything by 7?")
second_response = chat(messages)
second_response
```
