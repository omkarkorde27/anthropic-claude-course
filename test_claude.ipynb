```python
import os
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()

client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

def add_user_message(messages, text):
    user_message={"role":"user", "content":text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message={"role":"assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1024,
        messages=messages
    )

    return message.content[0].text
```

```python
# Making a list of messages
# This will store all our messages (from both user and assistant) and the context needed for LLM to reply to previous response
messages = []

# Adding initial user question to messages and passing it to the LLM
add_user_message(messages, "What is astrophysics in 2 sentences?")

messages

answer = chat(messages)
answer

# Take the response of claude(assistant) and add it back to messages
add_assistant_message(messages, answer)

# Once the first user message and assitant message is added. Provide a follow-up question using the add_user_message helper function
add_user_message(messages, "Explain it as if you are talking to a astophysicist.")

# Print the model's response and see if it could retrieve the context and asnwer the followup question accordingly
answer = chat(messages)
answer
```
